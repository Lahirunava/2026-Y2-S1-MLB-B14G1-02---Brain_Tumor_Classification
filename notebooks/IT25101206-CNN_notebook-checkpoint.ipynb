## 2. Data Ingestion & Inventory

Scan both `Training/` and `Testing/` splits, collect metadata for every image file found.

```python
SUPPORTED_EXTENSIONS = {'.jpg', '.jpeg', '.png', '.bmp', '.tif', '.tiff'}

def scan_dataset(root_dir: str, split_label: str) -> list:
    """Walk a split directory and collect (path, class_name, split) tuples."""
    records = []
    if not os.path.isdir(root_dir):
        print(f"  [WARN] Directory not found: {root_dir}")
        return records
    for class_name in sorted(os.listdir(root_dir)):
        class_dir = os.path.join(root_dir, class_name)
        if not os.path.isdir(class_dir):
            continue
        for fname in os.listdir(class_dir):
            ext = os.path.splitext(fname)[1].lower()
            if ext in SUPPORTED_EXTENSIONS:
                records.append({
                    'filepath'   : os.path.join(class_dir, fname),
                    'filename'   : fname,
                    'class_name' : class_name,
                    'split'      : split_label
                })
    return records

train_records = scan_dataset(TRAIN_DIR, 'train')
test_records  = scan_dataset(TEST_DIR,  'test')
all_records   = train_records + test_records

df_raw = pd.DataFrame(all_records)

print("=" * 55)
print("           DATASET INVENTORY REPORT")
print("=" * 55)
print(f"  Training images : {len(train_records):>6,}")
print(f"  Testing  images : {len(test_records):>6,}")
print(f"  Total    images : {len(all_records):>6,}")
print(f"  Classes found   : {df_raw['class_name'].nunique()}")
print("\n  Per-class counts (Training):")
print(df_raw[df_raw['split']=='train']['class_name']
      .value_counts()
      .to_string(header=False))
print("=" * 55)
```

## 3. Handling Missing / Corrupted Image Data

Every file is opened with Pillow. Images that raise an exception or cannot be decoded are flagged as **corrupted** and excluded from downstream processing.

```python
def probe_image(filepath: str) -> dict:
    """
    Attempt to open an image and read its metadata.
    Returns a dict with status, width, height, mode, and file size.
    """
    result = {
        'status'    : 'ok',
        'width'     : None,
        'height'    : None,
        'mode'      : None,
        'file_size' : os.path.getsize(filepath),
        'error_msg' : None
    }
    try:
        with Image.open(filepath) as img:
            img.verify()                    # detects truncated / corrupted files
        # Re-open after verify (verify() consumes the stream)
        with Image.open(filepath) as img:
            img.load()                      # force full decompression
            result['width']  = img.width
            result['height'] = img.height
            result['mode']   = img.mode
    except (UnidentifiedImageError, OSError, Exception) as e:
        result['status']    = 'corrupted'
        result['error_msg'] = str(e)
    return result

print("Probing all images for corruption / missing data …")
probed = []
for rec in tqdm(all_records, desc='Probing', unit='img'):
    meta = probe_image(rec['filepath'])
    probed.append({**rec, **meta})

df_probed = pd.DataFrame(probed)

# ── Summary ───────────────────────────────────────────────────────────────────
n_total     = len(df_probed)
n_ok        = (df_probed['status'] == 'ok').sum()
n_corrupted = (df_probed['status'] == 'corrupted').sum()

print("\n" + "=" * 55)
print("     CORRUPTION / MISSING DATA REPORT")
print("=" * 55)
print(f"  Total images scanned  : {n_total:>6,}")
print(f"  Valid images          : {n_ok:>6,}")
print(f"  Corrupted / unreadable: {n_corrupted:>6,}")
print(f"  Corruption rate       : {n_corrupted/n_total*100:.2f}%")

if n_corrupted > 0:
    print("\n  Corrupted files:")
    bad = df_probed[df_probed['status'] == 'corrupted'][['filepath', 'error_msg']]
    for _, row in bad.iterrows():
        print(f"    • {row['filepath']}")
        print(f"      Error: {row['error_msg']}")
else:
    print("\n  ✓ No corrupted or missing images detected.")
print("=" * 55)
```

```python
# Keep only valid images for all downstream steps
df_clean = df_probed[df_probed['status'] == 'ok'].reset_index(drop=True)

print(f"[Step 3 Complete] Clean dataset size: {len(df_clean):,} images")
print(df_clean[['class_name', 'split', 'width', 'height', 'mode']].head(5))
```

## 4. Categorical Encoding — Folder Names → Numeric Labels

The class folder names are string labels. We apply `sklearn.preprocessing.LabelEncoder` to produce a deterministic integer mapping that can be consumed by any downstream model.

```python
le = LabelEncoder()
df_clean['label'] = le.fit_transform(df_clean['class_name'])

# Build a readable mapping table
class_mapping = pd.DataFrame({
    'class_name'   : le.classes_,
    'numeric_label': le.transform(le.classes_)
})

print("=" * 45)
print("  CATEGORICAL ENCODING — LABEL MAP")
print("=" * 45)
print(class_mapping.to_string(index=False))
print("=" * 45)

# Validate
print("\n  Sample rows after encoding:")
print(df_clean[['filename', 'class_name', 'label', 'split']].sample(8, random_state=42).to_string(index=False))

# Persist mapping to results/outputs
mapping_path = os.path.join(OUTPUTS_DIR, 'class_label_mapping.csv')
class_mapping.to_csv(mapping_path, index=False)
print(f"\n  Mapping saved → {mapping_path}")
```

## 5. Outlier Removal — Abnormal Aspect Ratios

MRI scans from the same scanner should share a consistent aspect ratio (width ÷ height). Images with extreme or erratic aspect ratios are likely artefacts, mis-cropped scans, or data-entry errors and should be removed before training.

```python
# Compute aspect ratio
df_clean = df_clean.copy()
df_clean['aspect_ratio'] = df_clean['width'] / df_clean['height']

# IQR-based outlier bounds
Q1 = df_clean['aspect_ratio'].quantile(0.25)
Q3 = df_clean['aspect_ratio'].quantile(0.75)
IQR = Q3 - Q1
lower_bound = Q1 - 1.5 * IQR
upper_bound = Q3 + 1.5 * IQR

mask_outlier = ~df_clean['aspect_ratio'].between(lower_bound, upper_bound)
n_outliers   = mask_outlier.sum()

print("=" * 55)
print("          OUTLIER REMOVAL REPORT")
print("=" * 55)
print(f"  Aspect-ratio statistics:")
print(df_clean['aspect_ratio'].describe().rename(lambda x: f"    {x}").to_string())
print(f"\n  IQR          : {IQR:.4f}")
print(f"  Lower bound  : {lower_bound:.4f}")
print(f"  Upper bound  : {upper_bound:.4f}")
print(f"\n  Outlier images identified : {n_outliers}")

if n_outliers > 0:
    print("\n  Outlier details:")
    print(df_clean[mask_outlier][['filepath', 'width', 'height', 'aspect_ratio']].to_string(index=False))
else:
    print("  ✓ No aspect-ratio outliers detected.")

# Remove outliers
df_filtered = df_clean[~mask_outlier].reset_index(drop=True)
print(f"\n  Dataset after outlier removal: {len(df_filtered):,} images")
print("=" * 55)
```

## 6. Normalisation / Pixel Scaling (0 – 1)

Raw pixel values range **0 – 255** (uint8). Scaling to **[0, 1]** (float32) by dividing by 255 accelerates gradient-based optimisation and ensures consistent feature magnitudes across the dataset. We demonstrate and verify the transform on a sample of images.

```python
TARGET_SIZE   = (128, 128)   # used in normalisation sample AND later resizing
SAMPLE_SIZE   = min(200, len(df_filtered))

sample_df = df_filtered.sample(SAMPLE_SIZE, random_state=42).reset_index(drop=True)

pre_norm_stats  = {'min': [], 'max': [], 'mean': [], 'std': []}
post_norm_stats = {'min': [], 'max': [], 'mean': [], 'std': []}

print(f"Running normalisation verification on {SAMPLE_SIZE} sample images …")
for _, row in tqdm(sample_df.iterrows(), total=SAMPLE_SIZE, desc='Normalising', unit='img'):
    with Image.open(row['filepath']) as img:
        img_rgb  = img.convert('RGB').resize(TARGET_SIZE, Image.LANCZOS)
        arr_raw  = np.array(img_rgb, dtype=np.uint8)
        arr_norm = arr_raw.astype(np.float32) / 255.0

    pre_norm_stats['min'].append(arr_raw.min())
    pre_norm_stats['max'].append(arr_raw.max())
    pre_norm_stats['mean'].append(arr_raw.mean())
    pre_norm_stats['std'].append(arr_raw.std())

    post_norm_stats['min'].append(arr_norm.min())
    post_norm_stats['max'].append(arr_norm.max())
    post_norm_stats['mean'].append(arr_norm.mean())
    post_norm_stats['std'].append(arr_norm.std())

# Aggregate
print("\n" + "=" * 60)
print("        PIXEL NORMALISATION VERIFICATION REPORT")
print("=" * 60)
print(f"  Sample size : {SAMPLE_SIZE} images")
print(f"  Method      : divide by 255 (min-max scaling to [0, 1])")
print()
headers = f"  {'Statistic':<12} {'Pre-Norm (uint8)':>20} {'Post-Norm (float32)':>22}"
print(headers)
print("  " + "-" * (len(headers) - 2))
for stat in ['min', 'max', 'mean', 'std']:
    pre_val  = np.mean(pre_norm_stats[stat])
    post_val = np.mean(post_norm_stats[stat])
    print(f"  {stat.capitalize():<12} {pre_val:>20.4f} {post_val:>22.6f}")
print("=" * 60)
print(f"\n  ✓ All post-norm values within [0, 1]: ",
      all(v >= 0 and v <= 1 for v in post_norm_stats['min'] + post_norm_stats['max']))
```

## 7. Feature Engineering / Dimension Reduction — Resize to 128 × 128

Raw images have **heterogeneous spatial dimensions**. Resizing to a fixed 128 × 128 px:
- Standardises the input feature space
- Reduces memory and compute cost compared to native resolution
- Serves as the canonical pre-processing step before any CNN forward pass

We process **all** filtered images, verify the output shape, and save a summary CSV.

```python
TARGET_H, TARGET_W = 128, 128

print(f"Resizing all {len(df_filtered):,} images to {TARGET_H}×{TARGET_W} …")

shape_records = []   # to verify uniformity
failed_resize = []

for _, row in tqdm(df_filtered.iterrows(), total=len(df_filtered),
                   desc='Resizing', unit='img'):
    try:
        with Image.open(row['filepath']) as img:
            resized = img.convert('RGB').resize((TARGET_W, TARGET_H), Image.LANCZOS)
            arr     = np.array(resized, dtype=np.float32) / 255.0
        shape_records.append({'shape': arr.shape, 'dtype': str(arr.dtype)})
    except Exception as e:
        failed_resize.append({'filepath': row['filepath'], 'error': str(e)})

# Verification
unique_shapes = set(r['shape'] for r in shape_records)

print("\n" + "=" * 55)
print("       DIMENSION REDUCTION REPORT")
print("=" * 55)
print(f"  Target size          : {TARGET_H} × {TARGET_W} × 3 (RGB)")
print(f"  Images processed     : {len(shape_records):,}")
print(f"  Failed during resize : {len(failed_resize)}")
print(f"  Unique output shapes : {len(unique_shapes)}  → {unique_shapes}")
print(f"  Output dtype         : {shape_records[0]['dtype'] if shape_records else 'N/A'}")

if len(unique_shapes) == 1:
    print("\n  ✓ All images successfully standardised to a uniform shape.")
else:
    print("\n  ⚠ Multiple shapes detected — investigate further.")
print("=" * 55)

# Save processed metadata summary
df_filtered_out = df_filtered.copy()
df_filtered_out['resized_h']    = TARGET_H
df_filtered_out['resized_w']    = TARGET_W
df_filtered_out['resized_dtype']= 'float32'
summary_path = os.path.join(OUTPUTS_DIR, 'processed_metadata.csv')
df_filtered_out.to_csv(summary_path, index=False)
print(f"\n  Metadata CSV saved → {summary_path}")
```

### 8.1 — Class Distribution Bar Chart

```python
fig, axes = plt.subplots(1, 2, figsize=(14, 6))
fig.suptitle('Brain MRI — Class Distribution', fontsize=16, fontweight='bold', y=1.02)

for ax, split in zip(axes, ['train', 'test']):
    subset = df_filtered[df_filtered['split'] == split]
    counts = subset['class_name'].value_counts().reindex(list(CLASS_PALETTE.keys()), fill_value=0)
    colors = [CLASS_PALETTE.get(c, '#999999') for c in counts.index]

    bars = ax.bar(counts.index, counts.values, color=colors,
                  edgecolor='white', linewidth=1.5, alpha=0.9)

    # Annotate bars
    for bar, val in zip(bars, counts.values):
        ax.text(bar.get_x() + bar.get_width() / 2,
                bar.get_height() + max(counts.values) * 0.01,
                f'{val:,}', ha='center', va='bottom', fontsize=10, fontweight='bold')

    ax.set_title(f'{split.capitalize()} Split  (n={len(subset):,})', fontsize=13, pad=10)
    ax.set_xlabel('Tumor Class')
    ax.set_ylabel('Number of Images')
    ax.set_xticks(range(len(counts)))
    ax.set_xticklabels([c.replace('_', '\n') for c in counts.index], fontsize=9)
    ax.yaxis.grid(True, alpha=0.4, linestyle='--')
    ax.set_axisbelow(True)

plt.tight_layout()
save_path = os.path.join(EDA_VIS_DIR, '01_class_distribution.png')
fig.savefig(save_path, bbox_inches='tight')
plt.show()
print(f"  Plot saved → {save_path}")
```

### 8.2 — Pixel Intensity Histogram

```python
# Sample images per class for intensity analysis
INTENSITY_SAMPLE = 50   # images per class

intensity_data = {}   # class → flat pixel array

print(f"Sampling {INTENSITY_SAMPLE} images per class for intensity analysis …")
for class_name, color in CLASS_PALETTE.items():
    class_df  = df_filtered[
        (df_filtered['class_name'] == class_name) &
        (df_filtered['split'] == 'train')
    ].head(INTENSITY_SAMPLE)

    pixels = []
    for _, row in class_df.iterrows():
        try:
            with Image.open(row['filepath']) as img:
                arr = np.array(img.convert('L').resize((128, 128)), dtype=np.float32) / 255.0
            pixels.extend(arr.flatten().tolist())
        except Exception:
            pass
    intensity_data[class_name] = np.array(pixels)
    print(f"  {class_name:<22} — {len(pixels):,} pixels sampled")

# Plot
fig, axes = plt.subplots(2, 2, figsize=(14, 10))
fig.suptitle('Brain MRI — Pixel Intensity Distributions per Class\n'
             '(Grayscale, normalised to [0, 1])',
             fontsize=15, fontweight='bold')

for ax, (class_name, pixels) in zip(axes.flatten(), intensity_data.items()):
    color = CLASS_PALETTE[class_name]
    ax.hist(pixels, bins=80, color=color, alpha=0.80, edgecolor='white', linewidth=0.3)
    ax.axvline(np.mean(pixels), color='black', linestyle='--', linewidth=1.5,
               label=f'Mean={np.mean(pixels):.3f}')
    ax.axvline(np.median(pixels), color='grey', linestyle=':', linewidth=1.5,
               label=f'Median={np.median(pixels):.3f}')
    ax.set_title(class_name.replace('_', ' ').title(), fontsize=12)
    ax.set_xlabel('Normalised Pixel Intensity')
    ax.set_ylabel('Frequency')
    ax.legend(fontsize=9)
    ax.yaxis.grid(True, alpha=0.4, linestyle='--')
    ax.set_axisbelow(True)

plt.tight_layout()
save_path = os.path.join(EDA_VIS_DIR, '02_pixel_intensity_histogram.png')
fig.savefig(save_path, bbox_inches='tight')
plt.show()
print(f"  Plot saved → {save_path}")
```

### 8.3 — Image Dimension Scatter / Box Plot

```python
fig, axes = plt.subplots(1, 3, figsize=(18, 6))
fig.suptitle('Brain MRI — Raw Image Dimension Analysis', fontsize=16, fontweight='bold')

# ── Panel 1: Scatter plot (Width vs Height) ───────────────────────────────────
ax = axes[0]
for class_name, color in CLASS_PALETTE.items():
    subset = df_filtered[df_filtered['class_name'] == class_name]
    ax.scatter(subset['width'], subset['height'],
               c=color, alpha=0.4, s=12, label=class_name.replace('_', ' ').title())

ax.axhline(128, color='red', linestyle='--', linewidth=1, label='Target H=128')
ax.axvline(128, color='blue', linestyle='--', linewidth=1, label='Target W=128')
ax.set_xlabel('Image Width (px)')
ax.set_ylabel('Image Height (px)')
ax.set_title('Width vs Height — Scatter')
ax.legend(fontsize=7, markerscale=2)
ax.yaxis.grid(True, alpha=0.3)
ax.xaxis.grid(True, alpha=0.3)
ax.set_axisbelow(True)

# ── Panel 2: Box plot — Width distribution by class ───────────────────────────
ax = axes[1]
classes = list(CLASS_PALETTE.keys())
width_data = [df_filtered[df_filtered['class_name'] == c]['width'].values for c in classes]
bp = ax.boxplot(width_data, patch_artist=True, notch=False,
                medianprops=dict(color='black', linewidth=2))
for patch, (cls, color) in zip(bp['boxes'], CLASS_PALETTE.items()):
    patch.set_facecolor(color)
    patch.set_alpha(0.75)
ax.set_xticks(range(1, len(classes)+1))
ax.set_xticklabels([c.replace('_', '\n') for c in classes], fontsize=8)
ax.set_xlabel('Tumor Class')
ax.set_ylabel('Raw Image Width (px)')
ax.set_title('Width Distribution — Box Plot')
ax.yaxis.grid(True, alpha=0.4, linestyle='--')
ax.set_axisbelow(True)

# ── Panel 3: Box plot — Height distribution by class ──────────────────────────
ax = axes[2]
height_data = [df_filtered[df_filtered['class_name'] == c]['height'].values for c in classes]
bp2 = ax.boxplot(height_data, patch_artist=True, notch=False,
                 medianprops=dict(color='black', linewidth=2))
for patch, (cls, color) in zip(bp2['boxes'], CLASS_PALETTE.items()):
    patch.set_facecolor(color)
    patch.set_alpha(0.75)
ax.set_xticks(range(1, len(classes)+1))
ax.set_xticklabels([c.replace('_', '\n') for c in classes], fontsize=8)
ax.set_xlabel('Tumor Class')
ax.set_ylabel('Raw Image Height (px)')
ax.set_title('Height Distribution — Box Plot')
ax.yaxis.grid(True, alpha=0.4, linestyle='--')
ax.set_axisbelow(True)

plt.tight_layout()
save_path = os.path.join(EDA_VIS_DIR, '03_image_dimensions_boxplot.png')
fig.savefig(save_path, bbox_inches='tight')
plt.show()
print(f"  Plot saved → {save_path}")
```
